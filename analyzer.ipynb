```python
import numpy as np 

regioned = np.genfromtxt("nFlixCustomerChurn.csv",
                         delimiter=",", 
                         names=True,
                         dtype=None,
                         encoding="utf-8")

isMale = regioned["gender"] == "Male"
isFemale = regioned["gender"] == "Female"
isOthers = regioned["gender"] == "Other"

maleCust = regioned[isMale]
femaleCust = regioned[isFemale]
otherCust = regioned[isOthers]
regions = np.unique(regioned["region"])

genderMasks = [isFemale, isMale, isOthers]
genderNames = ["Female", "Male", "Other"]

counts = np.zeros((len(regions), len(genderMasks)), dtype=int)
for i, region in enumerate(regions):
    isRegion = regioned["region"] == region
    for j, gMask in enumerate(genderMasks):
        counts[i,j] = (isRegion & gMask).sum()

totals = counts.sum(axis=1)
print("Grid adds up to all customers: ", counts.sum() == len(regioned))
print("Gender columns match: ", np.array_equal(
    counts.sum(axis=0), [isFemale.sum(), isMale.sum(), isOthers.sum()]))

with open("region_render_counts.csv", "w") as f:
    f.write("region, " + ",".join(genderNames) + ",Total\n")
    for i, region in enumerate(regions):
        row = ",".join(str(c) for c in counts[i])
        f.write(f"{region}, {row}, {totals[i]}\n")

print(open("region_render_counts.csv").read())

```
