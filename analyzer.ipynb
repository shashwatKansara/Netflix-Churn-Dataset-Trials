```python
import numpy as np

data = np.genfromtxt("nFlixCustomerChurn.csv", 
                     delimiter=",",
                     names=True,
                     dtype=None,
                     encoding="utf-8")

num_accounts = len(data) #len provides a total size (rows) of the used dataset
profiles = data["number_of_profiles"]
total_profiles = profiles.sum()
print(num_accounts)
print("Total number of profiles across all accounts is: ", total_profiles)
```

```python
import numpy as np

numData = np.genfromtxt("nFlixCustomerChurn.csv",
                        delimiter=",",
                        names=True,
                        dtype=None,
                        encoding="utf-8")

hours = numData["watch_hours"]
print("Max Hours watched: ", hours.max())
print("Minimum Hours watched: ", hours.min())

# additional trials:
userAge = numData["age"]
print("eldest viewer: ", userAge.max())
print("youngest viewer: ", userAge.min())
```

```python
import numpy as np 

userData = np.genfromtxt("nFlixCustomerChurn.csv", 
                         delimiter=",",
                         names=True,
                         dtype=None,
                         encoding="utf-8")

# boolean indexing - first line ensures filters male customers as boolean and second line consists
# of only male customers and dropping non-male customers. 

isMale = userData["gender"] == "Male"
maleCustomers = numData[isMale]
totalMaleCustomers = len(maleCustomers)
print("Total Male customers are: ", totalMaleCustomers)

# same for female and 'other' gender category customers

isFemale = userData["gender"] == "Female"
femaleCustomers = userData[isFemale]
totalFemaleCustomers = len(femaleCustomers)
print("Total Female customers are: ", totalFemaleCustomers)

otherGender = userData["gender"] == "Other"
other = userData[otherGender]
totalOtherGender = len(other)
print("Total customers with 'other' as their genders are: ", totalOtherGender)


```
